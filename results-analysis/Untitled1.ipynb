```python
import pandas as pd
df_demog = pd.read_csv("prolific_v3_demog.csv")
df_demog = df_demog.rename(columns={'Participant id':'participant_name'})
df_participants = pd.read_csv("participants.csv")
df_participants['participant_name'] = df_participants['participant_name'].apply(lambda row: row[len('v3_prolific -- '):])
df_participants = df_participants.drop(columns=['Unnamed: 0'])
df_participants = df_participants.drop_duplicates()
df_demog.head()
```

```python
data = pd.merge(df_demog, df_participants, on='participant_name')
zpdes, baseline = data[data['condition']=='zpdes'], data[data['condition']=='baseline']
```

```python
from pingouin import ttest
ttest(zpdes['Age'], baseline['Age'], paired=False)
```

```python
ttest(zpdes['Total approvals'], baseline['Total approvals'], paired=False)
```

```python
from scipy.stats import chi2_contingency
contingency_table = pd.crosstab(data['condition'], data['Sex'])

# Perform the chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Output the results
print(f"Chi-square value: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
print(f"Expected counts: \n{expected}")
```
